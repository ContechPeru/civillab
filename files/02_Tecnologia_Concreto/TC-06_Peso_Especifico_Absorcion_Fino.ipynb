## 🧠 ¿De dónde sale el denominador $(B + S - C)$ del Picnómetro?
- Si pusieras el matraz lleno de agua sola (**$B$**) y la arena SSS (**$S$**) juntos en una balanza, pesarían **$B + S$**.
- Al introducir los $S$ gramos de arena en el matraz lleno y volver a enrasar a la misma marca de $500\text{ cm}^3$ (**$C$**), la arena obligó a desbordar un volumen de agua idéntico al volumen de sus granos.
- Esa agua desalojada vale exactamente: **$(B + S) - C = B + S - C$** ($\text{cm}^3$).
