## 🧠 ¿De dónde sale la división por $(B - C)$ y $(A - C)$?
- Por el **Principio de Arquímedes**, cuando sumerges la grava saturada (que en aire pesa **$B$**) en la canastilla del agua (pesada como **$C$**), el agua empuja hacia arriba reduciendo su peso exactamente en el **peso del agua desalojada = $(B - C)$**. Como $1\text{ g de agua} = 1\text{ cm}^3$, entonces **$(B - C)$ es el volumen total de los granos (con poros)** en $\text{cm}^3$.
- Si al volumen total le restas el agua que se alojó en los poros $(B - A)$, queda el volumen sólido impermeable: $(B - C) - (B - A) = \mathbf{A - C}$.
