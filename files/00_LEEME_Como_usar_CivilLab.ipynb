# CivilLab UNSAAC v1.0.0 — versión web
**Facultad de Ingeniería Civil — Laboratorio de Materiales y Concreto**  
Tecnología de Materiales (IC151) · Tecnología del Concreto (IC152)  
Autor y desarrollador: Ing. Jorge Luis Dueñas Ttito

Esta versión funciona **dentro del navegador**: no hay que instalar Python ni ningún programa.

## Cómo usarlo
1. En el panel de la izquierda, abra la carpeta de su curso:
   - `01_Tecnologia_Materiales`: prácticas TM-01 a TM-12
   - `02_Tecnologia_Concreto`: prácticas TC-01 a TC-11
   - `03_Complementarios`: DM-01, diseño de mezcla ACI 211.1
2. Haga doble clic en la práctica.
3. En el menú, elija **Ejecutar → Ejecutar todas las celdas** (*Run → Run All Cells*).
4. La primera vez tarda entre 30 y 60 segundos mientras el navegador carga el motor de cálculo. Después es inmediato.
5. Para usar sus propios datos, cambie los valores de la celda de datos de la hoja de laboratorio y vuelva a ejecutar.
6. Las fichas en Excel aparecen en la carpeta `fichas_generadas`. Para bajarlas: clic derecho sobre el archivo → **Descargar**.

## Importante
- Los cambios se guardan **solo en este navegador**. Descargue sus fichas antes de cerrar o cambiar de equipo.
- Use Chrome, Edge o Firefox actualizados.
- Para volver los cuadernos a su estado original: clic derecho sobre el archivo → **Eliminar**; el original vuelve a aparecer al recargar la página.
