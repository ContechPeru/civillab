## 🧠 ¿Por qué son esas divisiones? (Física del Ensayo)
En la canastilla hidrostática medimos el peso en aire seco (**$A$**), en aire saturado superficialmente seco (**$B$**) y sumergido en agua (**$C$**):
- **Principio de Arquímedes:** La pérdida aparente de peso al sumergir el agregado saturado es igual al peso del agua desplazada. Como $1\text{ g de agua} = 1\text{ cm}^3$, el **volumen total de la piedra con sus poros** vale exactamente **$(B - C)$**.
  - $G_{OD} = \dfrac{A}{B - C}$ &nbsp;|&nbsp; $G_{SSS} = \dfrac{B}{B - C}$
- Si restamos al volumen total el agua que entró en los poros $(B - A)$, obtenemos el **volumen sólido impermeable**: $(B - C) - (B - A) = \mathbf{A - C}$. De ahí nace el aparente: $G_a = \dfrac{A}{A - C}$.
