## 🧠 ¿Por qué el denominador es $(B + S - C)$ en el Picnómetro?
- Imagina el frasco lleno de agua sola hasta la marca (**$B$**) y en la otra mano la muestra de arena SSS (**$S$**, ≈ $500\text{ g}$). Juntos pesan **$B + S$**.
- Al meter la arena al frasco y enrasar de nuevo a la marca (**$C$**), la arena expulsó fuera del frasco un volumen de agua idéntico al volumen de sus granos.
- Esa agua expulsada es la diferencia de masas: **$(B + S) - C = B + S - C$** ($\text{cm}^3$).
- Al restarle el agua dentro de los poros $(S - A)$, queda el volumen sólido impermeable: **$(B + A - C)$**.
